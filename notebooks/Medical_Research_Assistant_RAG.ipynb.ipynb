```python
text = ""

for page in reader.pages:
    page_text = page.extract_text()

    if page_text:
        text += page_text + "\n"

print("Characters extracted:", len(text))
```

```python
import re

clean_text = re.sub(r'\s+', ' ', text).strip()

print("Original characters:", len(text))
print("Cleaned characters:", len(clean_text))
print(clean_text[:2000])
```

```python
chunk_size = 1000
chunk_overlap = 200

chunks = []

start = 0

while start < len(clean_text):
    end = start + chunk_size
    chunk = clean_text[start:end]
    chunks.append(chunk)

    start += chunk_size - chunk_overlap

print("Number of chunks:", len(chunks))
print("First chunk:\n")
print(chunks[0])
```

```python
print("First chunk length:", len(chunks[0]))
print("Second chunk length:", len(chunks[1]))

print("\n--- CHUNK 1 ---")
print(chunks[0])

print("\n--- CHUNK 2 ---")
print(chunks[1])
```

```python
print("Last chunk length:", len(chunks[-1]))
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded!")
```

```python
test_embedding = embedding_model.encode(chunks[0])

print("Embedding type:", type(test_embedding))
print("Embedding dimensions:", test_embedding.shape)
```

```python
embeddings = embedding_model.encode(
    chunks,
    show_progress_bar=True
)

print("Number of embeddings:", len(embeddings))
print("Embedding dimensions:", embeddings.shape)
```

```python
import chromadb

CHROMA_PATH = os.path.join(PROJECT_DIR, "chroma_db")

client = chromadb.PersistentClient(path=CHROMA_PATH)

print("ChromaDB client created successfully!")
```

```python
collection = client.get_or_create_collection(
    name="medical_documents"
)

print("Collection created successfully!")
```

```python
ids = [f"chunk_{i}" for i in range(len(chunks))]

print("Number of IDs:", len(ids))
print("First 5 IDs:", ids[:5])
```

```python
collection.add(
    ids=ids,
    documents=chunks,
    embeddings=embeddings.tolist()
)

print("Medical document chunks added to ChromaDB!")
```

```python
print("Chunks:", len(chunks))
print("ChromaDB documents:", collection.count())
```

```python
query = "What are the limitations of RAG systems in handling noise and misinformation?"
```

```python
query_embedding = embedding_model.encode(query).tolist()

print("Query embedding dimensions:", len(query_embedding))
```

```python
results = collection.query(
    query_embeddings=[query_embedding],
    n_results=3
)
```

```python
for i, document in enumerate(results["documents"][0]):
    print(f"\n--- Result {i+1} ---")
    print(document)
```

```python
results = collection.query(
    query_embeddings=[query_embedding],
    n_results=3
)
```

```python
retrieved_chunks = results["documents"][0]

print("Number of retrieved chunks:", len(retrieved_chunks))
```

```python
context = "\n\n".join(retrieved_chunks)

print(context)
```
